```python
#4 Xoay ảnh xoay 45 độ với reshape =True và reshape =Fale và sau đó nhận xét
import numpy as np
import scipy.ndimage as nd
import imageio.v2 as iio
import matplotlib.pylab as plt

# Đọc ảnh 'fruit.jpg'
data = iio.imread('fruit.jpg')

# In kích thước của ảnh gốc
print(f"Kích thước ảnh gốc: {data.shape}")

# Xoay ảnh 45 độ với reshape=True
rotated_data_reshape_true = nd.rotate(data, 45, reshape=True)

# Xoay ảnh 45 độ với reshape=False
rotated_data_reshape_false = nd.rotate(data, 45, reshape=False)

# Hiển thị ảnh đã xoay với reshape=True
plt.subplot(1, 2, 1)
plt.imshow(rotated_data_reshape_true)
plt.title("Xoay 45 độ với reshape=True")

# Hiển thị ảnh đã xoay với reshape=False
plt.subplot(1, 2, 2)
plt.imshow(rotated_data_reshape_false)
plt.title("Xoay 45 độ với reshape=False")

plt.show()

# In kích thước của ảnh sau khi xoay
print(f"Kích thước ảnh sau khi xoay với reshape=True: {rotated_data_reshape_true.shape}")
print(f"Kích thước ảnh sau khi xoay với reshape=False: {rotated_data_reshape_false.shape}")
```

Nhận xét:
reshape=True tạo ra một ảnh lớn hơn (kích thước tăng lên) với không gian trống xung quanh để chứa toàn bộ hình ảnh sau khi xoay, điều này giữ toàn bộ ảnh.

reshape=False giữ kích thước ảnh ban đầu, nhưng có thể bị cắt bớt nếu phần ảnh vượt ra ngoài khung ban đầu.



```python
#5 Dialtion và Erosion (Thay đổi kích thước size=(5, 5) và quan sát sự khác biệt sau đó nhận xét
import numpy as np
import scipy.ndimage as nd
import imageio.v2 as iio
import matplotlib.pylab as plt

# Đọc ảnh 'world_cup.jpg' và chuyển thành ảnh grayscale
data = iio.imread('world_cup.jpg', mode='L')

# In kích thước của ảnh
print(data.shape)

# Thực hiện dilation (nở ảnh) một lần và hiển thị
d1 = nd.binary_dilation(data)
plt.imshow(d1)
plt.show()

# Thực hiện dilation 3 lần và hiển thị
d2 = nd.binary_dilation(data, iterations=3)
plt.imshow(d2)
plt.show()
```

```python
import numpy as np
import scipy.ndimage as nd
import imageio.v2 as iio
import matplotlib.pyplot as plt

# Định nghĩa hàm GeoFun với hàm sin thay vì cos
def GeoFun(outcoord):
    # Thay cos bằng sin để thay đổi hình dạng
    a = 10 * np.sin(outcoord[0]/10.0) + outcoord[0]
    b = 10 * np.sin(outcoord[1]/10.0) + outcoord[1]
    return a, b

# Đọc ảnh với mode='L' để ảnh xám (grayscale)
data = iio.imread('world_cup.jpg', mode='L')

# Áp dụng biến đổi hình học với hàm GeoFun đã thay đổi
d1 = nd.geometric_transform(data, GeoFun)
plt.imshow(d1, cmap='gray')
plt.title("Biến đổi với GeoFun thay đổi (sin)")
plt.show()

# Áp dụng phép toán Dilation với kernel (5, 5)
dilated = nd.binary_dilation(data, structure=np.ones((5, 5)))
plt.imshow(dilated, cmap='gray')
plt.title("Dilation với kernel (5, 5)")
plt.show()

# Áp dụng phép toán Erosion với kernel (5, 5)
eroded = nd.binary_erosion(data, structure=np.ones((5, 5)))
plt.imshow(eroded, cmap='gray')
plt.title("Erosion với kernel (5, 5)")
plt.show()
```

Nhận xét: Biến đổi hình học với hàm sin tạo ra các hình dạng sóng mạnh mẽ, làm cho ảnh gốc bị méo và biến dạng.

Dilation mở rộng các vùng sáng trong ảnh, làm cho chúng trở nên lớn và rõ rệt hơn, nhưng đồng thời làm mất các chi tiết nhỏ.

Erosion làm thu nhỏ các vùng sáng trong ảnh, làm cho các chi tiết nhỏ biến mất, chỉ còn lại những vùng sáng lớn và dễ nhận diện.

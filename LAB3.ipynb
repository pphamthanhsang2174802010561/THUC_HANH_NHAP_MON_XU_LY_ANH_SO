```python
#1.6 Coorfinate Mapping
import numpy as np
import scipy.ndimage as nd
import imageio.v2 as iio
import matplotlib.pylab as plt

# Đọc ảnh 'world_cup.jpg' và chuyển thành ảnh grayscale
data = iio.imread('world_cup.jpg', mode='L')

# In kích thước của ảnh
print(data.shape)

# Lấy chiều cao và chiều rộng của ảnh
V, H = data.shape

# Tạo chỉ số cho các điểm của ảnh
m = np.indices((V, H))

# Khoảng dịch chuyển
d = 5
q = 2 * d * np.random.rand(m.shape[1], m.shape[2]) - d

# Dịch chuyển các điểm của ảnh
mp = (m + q).astype(int)

# Áp dụng dịch chuyển vào ảnh
d1 = nd.map_coordinates(data, mp)

# Hiển thị ảnh đã dịch chuyển
plt.imshow(d1)
plt.show()
```

```python
#1.7 Biến đổi chung
import numpy as np
import scipy.ndimage as nd
import imageio.v2 as iio
import matplotlib.pylab as plt

# Hàm GeoFun để tính toán dịch chuyển các pixel
def GeoFun(outcoord):
    a = 10 * np.cos(outcoord[0] / 10.0) + outcoord[0]
    b = 10 * np.cos(outcoord[1] / 10.0) + outcoord[1]
    return a, b

# Đọc ảnh 'world_cup.jpg' và chuyển thành ảnh grayscale
data = iio.imread('world_cup.jpg', mode='L')  # Sử dụng 'mode=L' thay vì 'as_gray=True'

# Áp dụng phép biến đổi hình học
dl = nd.geometric_transform(data, GeoFun)

# Hiển thị ảnh đã biến đổi
plt.imshow(dl, cmap='gray')  # Dùng cmap='gray' để hiển thị ảnh grayscale
plt.show()
```

```python
#Câu 3:
import cv2
import numpy as np

# Khai báo các biến toàn cục
drawing = False  # Biến để kiểm tra xem có đang vẽ hay không
ix, iy = -1, -1  # Tọa độ bắt đầu khi nhấn chuột
roi = None  # Vùng chọn

# Hàm xử lý khi người dùng khoanh vùng ảnh
def select_roi(event, x, y, flags, param):
    global ix, iy, drawing, img, roi

    # Khi nhấn chuột trái
    if event == cv2.EVENT_LBUTTONDOWN:
        drawing = True
        ix, iy = x, y

    # Khi kéo chuột trái
    elif event == cv2.EVENT_MOUSEMOVE:
        if drawing:
            img_copy = img.copy()  # Tạo bản sao của ảnh để vẽ
            cv2.rectangle(img_copy, (ix, iy), (x, y), (0, 255, 0), 2)  # Vẽ hình chữ nhật
            cv2.imshow("Image", img_copy)

    # Khi nhả chuột trái
    elif event == cv2.EVENT_LBUTTONUP:
        drawing = False
        roi = (ix, iy, x, y)  # Lưu lại tọa độ vùng chọn
        cv2.rectangle(img, (ix, iy), (x, y), (0, 255, 0), 2)  # Vẽ hình chữ nhật
        cv2.imshow("Image", img)
        print(f"Tọa độ vùng chọn: ({ix}, {iy}) đến ({x}, {y})")

# Đọc ảnh
img = cv2.imread('pagoda.jpg')

# Tạo cửa sổ và gọi hàm xử lý khi chọn vùng
cv2.imshow("Image", img)
cv2.setMouseCallback("Image", select_roi)

# Đợi người dùng nhấn phím bất kỳ để thoát
cv2.waitKey(0)
cv2.destroyAllWindows()

# Lấy tọa độ của vùng chọn
if roi:
    x1, y1, x2, y2 = roi
    kiwi = img[y1:y2, x1:x2]  # Cắt vùng chứa quả kiwi

    # Hiển thị ảnh quả kiwi đã chọn
    cv2.imshow("Kiwi", kiwi)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
#Câu 4
import numpy as np
import cv2  # OpenCV để thao tác với ảnh
import imageio.v2 as iio
import matplotlib.pylab as plt

# Đọc ảnh từ file 'quang_ninh.jpg'
image = iio.imread('quang_ninh.jpg')

# Tọa độ ngọn núi (417, 22) đến (661, 329)
x1_m, y1_m, x2_m, y2_m = 417, 22, 661, 329
mountain = image[y1_m:y2_m, x1_m:x2_m]

# Tọa độ con thuyền (491, 457) đến (657, 543)
x1_b, y1_b, x2_b, y2_b = 491, 457, 657, 543
boat = image[y1_b:y2_b, x1_b:x2_b]

# Hàm xoay ảnh
def rotate_image(image, angle):
    # Lấy chiều cao và chiều rộng của ảnh
    (h, w) = image.shape[:2]
    # Tạo ma trận xoay quanh trung tâm ảnh
    center = (w // 2, h // 2)
    rotation_matrix = cv2.getRotationMatrix2D(center, angle, 1.0)  # Xoay theo góc 'angle'
    # Xoay ảnh
    rotated_image = cv2.warpAffine(image, rotation_matrix, (w, h))
    return rotated_image

# Xoay ngọn núi 45 độ
rotated_mountain = rotate_image(mountain, 45)

# Xoay con thuyền 45 độ
rotated_boat = rotate_image(boat, 45)

# Lưu ảnh đã xoay vào máy
cv2.imwrite('rotated_mountain.jpg', rotated_mountain)
cv2.imwrite('rotated_boat.jpg', rotated_boat)

# Hiển thị kết quả
plt.subplot(1, 2, 1)
plt.imshow(rotated_mountain)
plt.title("Rotated Mountain")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(rotated_boat)
plt.title("Rotated Boat")
plt.axis('off')

plt.show()
```

```python
#Cấu 5
import numpy as np
import cv2  # OpenCV để thao tác với ảnh
import imageio.v2 as iio
import matplotlib.pylab as plt

# Đọc ảnh từ file 'pagoda.jpg'
image = iio.imread('pagoda.jpg')

# Tọa độ ngôi chùa (bạn cung cấp tọa độ này)
x1, y1, x2, y2 = 54, 128, 505, 270  # Tọa độ ngôi chùa (thay đổi theo ảnh của bạn)

# Cắt ngôi chùa từ ảnh
pagoda = image[y1:y2, x1:x2]

# Tăng kích thước ngôi chùa lên 5 lần
height, width = pagoda.shape[:2]
resized_pagoda = cv2.resize(pagoda, (width * 5, height * 5), interpolation=cv2.INTER_LINEAR)

# Lưu ảnh ngôi chùa đã thay đổi kích thước
cv2.imwrite('resized_pagoda.jpg', resized_pagoda)

# Hiển thị kết quả
plt.imshow(resized_pagoda)
plt.title("Resized Pagoda (5x)")
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import imageio.v2 as iio
import matplotlib.pyplot as plt

# Hàm tịnh tiến (di chuyển ảnh)
def translate_image(image, tx, ty):
    translation_matrix = np.float32([[1, 0, tx], [0, 1, ty]])
    translated_image = cv2.warpAffine(image, translation_matrix, (image.shape[1], image.shape[0]))
    return translated_image

# Hàm xoay ảnh
def rotate_image(image, angle):
    (h, w) = image.shape[:2]
    center = (w // 2, h // 2)
    rotation_matrix = cv2.getRotationMatrix2D(center, angle, 1.0)
    rotated_image = cv2.warpAffine(image, rotation_matrix, (w, h))
    return rotated_image

# Hàm phóng to (resize ảnh)
def resize_image(image, scale_factor):
    height, width = image.shape[:2]
    resized_image = cv2.resize(image, (int(width * scale_factor), int(height * scale_factor)))
    return resized_image

# Hàm thu nhỏ (resize ảnh)
def shrink_image(image, scale_factor):
    height, width = image.shape[:2]
    resized_image = cv2.resize(image, (int(width * scale_factor), int(height * scale_factor)))
    return resized_image

# Hàm ánh xạ tọa độ (Coordinate Map)
def coordinate_map(image):
    rows, cols = image.shape[:2]
    map_x, map_y = np.meshgrid(np.arange(cols), np.arange(rows))
    map_x = np.float32(map_x)
    map_y = np.float32(map_y)
    return cv2.remap(image, map_x, map_y, interpolation=cv2.INTER_LINEAR)

# Hàm tạo menu
def show_menu():
    print("\nChọn phép biến đổi ảnh:")
    print("T - Tịnh tiến")
    print("X - Xoay")
    print("P - Phóng to")
    print("H - Thu nhỏ")
    print("C - Coordinate Map")
    print("Q - Thoát")

# Hàm chính
def main():
    # Đọc ba ảnh từ thư mục exercise (hoặc thay đổi đường dẫn nếu cần)
    image_paths = ["exercise/1.jpg", "exercise/2.jpg", "exercise/3.jpg"]
    
    # Hiển thị menu và yêu cầu người dùng chọn phép biến đổi
    while True:
        show_menu()
        choice = input("Nhập lựa chọn của bạn: ").upper()
        
        if choice == 'Q':
            print("Thoát chương trình.")
            break
        
        # Yêu cầu chọn ảnh
        print("Chọn ảnh để áp dụng phép biến đổi:")
        for i, path in enumerate(image_paths, 1):
            print(f"{i} - {path}")
        image_choice = int(input("Nhập số tương ứng với ảnh bạn muốn chọn: ")) - 1
        
        if image_choice < 0 or image_choice >= len(image_paths):
            print("Lựa chọn không hợp lệ. Vui lòng thử lại.")
            continue
        
        image = iio.imread(image_paths[image_choice])  # Đọc ảnh đã chọn
        print(f"Đang xử lý ảnh: {image_paths[image_choice]}")

        # Thực hiện phép biến đổi dựa trên lựa chọn của người dùng
        if choice == 'T':
            tx = int(input("Nhập độ dịch chuyển theo chiều ngang (pixels): "))
            ty = int(input("Nhập độ dịch chuyển theo chiều dọc (pixels): "))
            transformed_image = translate_image(image, tx, ty)
        
        elif choice == 'X':
            angle = float(input("Nhập góc xoay (độ): "))
            transformed_image = rotate_image(image, angle)
        
        elif choice == 'P':
            scale_factor = float(input("Nhập tỷ lệ phóng to (ví dụ 1.5 cho phóng to 1.5 lần): "))
            transformed_image = resize_image(image, scale_factor)
        
        elif choice == 'H':
            scale_factor = float(input("Nhập tỷ lệ thu nhỏ (ví dụ 0.5 cho thu nhỏ 1/2): "))
            transformed_image = shrink_image(image, scale_factor)
        
        elif choice == 'C':
            transformed_image = coordinate_map(image)
        
        # Hiển thị kết quả sau phép biến đổi
        plt.imshow(transformed_image)
        plt.title(f"Transformed Image ({choice})")
        plt.axis('off')
        plt.show()

        # Lưu kết quả vào máy
        output_path = f"transformed_{image_paths[image_choice].split('/')[-1]}"
        cv2.imwrite(output_path, transformed_image)
        print(f"Ảnh đã được lưu tại: {output_path}")

if __name__ == "__main__":
    main()
```
